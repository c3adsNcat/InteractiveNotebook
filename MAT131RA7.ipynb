# Part 1 – Section 7.3: Concavity and the Second Derivative Test

## 1. Concavity and Inflection Points

Let $f$ be differentiable on an interval.

- $f$ is **concave up** if $f'$ is **increasing** on the interval. A test: if $f''(x) > 0$, the graph is concave up (shaped like a cup ∪).
- $f$ is **concave down** if $f'$ is **decreasing** on the interval. A test: if $f''(x) < 0$, the graph is concave down (shaped like a cap ∩).
- An **inflection point** is a point on the graph where the **concavity changes**. Candidates are where $f''(x) = 0$ or $f''$ is undefined, but you must check that $f''$ actually **changes sign**.



#### 🖼️ Image description
Three graphs side by side.
1. **Concave up** (blue): a U-shaped curve with dashed tangent lines; the tangent slopes increase from left to right and the curve lies **above** its tangents.
2. **Concave down** (orange): an upside-down U with dashed tangent lines; the slopes decrease and the curve lies **below** its tangents.
3. **Inflection point**: the graph of $f(x) = x^3 - 3x^2$ is orange (concave down) to the left of $x = 1$ and blue (concave up) to the right. A green dot marks the inflection point $(1, -2)$, where the concavity changes.

### Example 1 (Concavity and Inflection Points)

Find the intervals of concavity and the inflection point of $f(x) = x^3 - 3x^2$.

$$f'(x) = 3x^2 - 6x \qquad\qquad f''(x) = 6x - 6$$

Set $f''(x) = 0$: $\;6x - 6 = 0 \Rightarrow x = 1$. Test a value on each side:

| Interval | Test value | Sign of $f''$ | Concavity |
|---|---|---|---|
| $(-\infty, 1)$ | $x = 0$ | $f''(0) = -6 < 0$ | concave down |
| $(1, \infty)$ | $x = 2$ | $f''(2) = 6 > 0$ | concave up |

Since $f''$ changes sign at $x = 1$, there is an inflection point at $\big(1, f(1)\big) = (1, -2)$.

**Explanation**

$f''(x) = 0$ only gives a **candidate**. For example, $f(x) = x^4$ has $f''(0) = 0$, but $f'' = 12x^2 \ge 0$ never changes sign, so $x^4$ has **no** inflection point.

## 2. The Second Derivative Test

Suppose $f'(c) = 0$ (so $c$ is a critical number) and $f''$ exists near $c$.

| If… | then $f$ has… | Picture |
|---|---|---|
| $f''(c) > 0$ | a **local minimum** at $c$ | concave up, like the bottom of a cup ∪ |
| $f''(c) < 0$ | a **local maximum** at $c$ | concave down, like the top of a cap ∩ |
| $f''(c) = 0$ | **the test fails**: use the First Derivative Test instead | — |

### Example 2

Classify the critical numbers of $f(x) = x^3 - 3x$.

$$f'(x) = 3x^2 - 3 = 3(x-1)(x+1) = 0 \;\Rightarrow\; x = -1,\ x = 1 \qquad\qquad f''(x) = 6x$$

- $f''(-1) = -6 < 0 \;\Rightarrow\;$ **local maximum** at $x = -1$, value $f(-1) = 2$.
- $f''(1) = 6 > 0 \;\Rightarrow\;$ **local minimum** at $x = 1$, value $f(1) = -2$.

**Explanation**

The Second Derivative Test is often faster than a sign chart for $f'$, because you only plug each critical number into $f''$. But when $f''(c) = 0$ it gives **no** information: $x^4$, $-x^4$, and $x^3$ all have $f'(0) = f''(0) = 0$, yet they have a minimum, a maximum, and neither at $x = 0$.

# Part 2 – Section 7.4: Graphing Functions

## 3. A Strategy for Graphing

To sketch the graph of $y = f(x)$, gather the following information:

1. **Domain:** where is $f$ defined?
2. **Intercepts:** the $y$-intercept $f(0)$, and $x$-intercepts where $f(x) = 0$ (if easy to find).
3. **Symmetry:** is $f$ even ($f(-x) = f(x)$) or odd ($f(-x) = -f(x)$)?
4. **Asymptotes:** vertical asymptotes (where the denominator is $0$) and horizontal asymptotes (limits as $x \to \pm\infty$), for rational functions.
5. **First derivative:** critical numbers, intervals of **increase/decrease**, local **max/min**.
6. **Second derivative:** intervals of **concavity** and **inflection points**.
7. **Plot** the key points and connect them, following the shapes found in steps 5 and 6.

### Example 3 (Putting It All Together)

Sketch $f(x) = x^3 - 3x^2 - 9x + 5$.

**Domain:** all real numbers. **$y$-intercept:** $f(0) = 5$. No asymptotes (it's a polynomial).

**First derivative:** $f'(x) = 3x^2 - 6x - 9 = 3(x-3)(x+1)$, so the critical numbers are $x = -1$ and $x = 3$.

| Interval | $(-\infty, -1)$ | $(-1, 3)$ | $(3, \infty)$ |
|---|---|---|---|
| Sign of $f'$ | $+$ | $-$ | $+$ |
| $f$ is | increasing | decreasing | increasing |

So $f$ has a **local max** at $(-1, 10)$ and a **local min** at $(3, -22)$.

**Second derivative:** $f''(x) = 6x - 6 = 0 \Rightarrow x = 1$.

| Interval | $(-\infty, 1)$ | $(1, \infty)$ |
|---|---|---|
| Sign of $f''$ | $-$ | $+$ |
| $f$ is | concave down | concave up |

So there is an **inflection point** at $(1, -6)$.



#### 🖼️ Image description
The graph of $f(x) = x^3 - 3x^2 - 9x + 5$ for $x$ from $-3$ to $5$. The background is shaded **orange** to the left of $x = 1$ (labeled "concave down, $f''(x) < 0$") and **blue** to the right of $x = 1$ (labeled "concave up, $f''(x) > 0$"). The curve rises to a **local maximum** at $(-1, 10)$ (orange dot), passes through the **$y$-intercept** $(0, 5)$ (black dot), falls through the **inflection point** $(1, -6)$ (green dot), reaches a **local minimum** at $(3, -22)$ (blue dot), and then rises again.

**Explanation**

The sign charts tell you the **shape** between the key points: increasing and concave down near the max, decreasing while switching from concave down to up, and increasing and concave up after the min.

# Common Mistakes

- Assuming every $x$ with $f''(x) = 0$ is an inflection point. You must check that $f''$ **changes sign**.
- Mixing up the Second Derivative Test: $f''(c) > 0$ means **minimum** (cup ∪), $f''(c) < 0$ means **maximum** (cap ∩).
- Concluding "no extremum" when $f''(c) = 0$. The test simply **fails**; use the First Derivative Test.
- Calling every critical number an extremum, like $x = 0$ for $x^4 - 4x^3$, where $f'$ does not change sign.
- Writing coordinates with only $x$: an inflection point or extremum is a **point** $(c, f(c))$.

# Connecting to Data Science

Concavity describes whether a trend is **speeding up** or **slowing down**, which data scientists care about a lot:

- **Growth that is slowing down.** A new app's user count may still be increasing ($f' > 0$) while growth slows ($f'' < 0$). Spotting this concave-down shape early helps companies plan before growth levels off.
- **Inflection points in S-shaped curves.** Product adoption and the spread of a virus often follow an S-shaped curve. The inflection point is where growth is **fastest**; after it, growth begins to slow. Analysts watch for it to predict the peak.
- **Machine learning finds minimums.** Training a model means finding the parameters that **minimize** its error. When the error curve is **concave up** (a "bowl"), the Second Derivative Test guarantees that the bottom is a true minimum, which is why data scientists prefer models with bowl-shaped error.

# Summary

- $f'' > 0$ means **concave up** (∪); $f'' < 0$ means **concave down** (∩). An **inflection point** is where the concavity changes.
- **Second Derivative Test:** at a critical number $c$, $f''(c) > 0$ gives a local **min**, $f''(c) < 0$ a local **max**, and $f''(c) = 0$ gives no conclusion.
- To **graph** a function, combine the domain, intercepts, symmetry, asymptotes, the sign of $f'$ (increasing/decreasing), and the sign of $f''$ (concavity).
