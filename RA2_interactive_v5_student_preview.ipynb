## 6.1 Composite Functions

Given two functions $f$ and $g$, the **composite function** $(f \circ g)(x)$ is defined by

$$
(f \circ g)(x) = f(g(x)).
$$

This means:
1. Start with an input $x$,
2. Apply $g$ to get $g(x)$,
3. Then apply $f$ to that result.

---

### Example 1: Forming a Composite

Let
$$
f(x) = 2x + 1, \qquad g(x) = x^2.
$$

Compute $(f \circ g)(x)$:

$$
(f \circ g)(x) = f(g(x)) = f(x^2) = 2x^2 + 1.
$$

---

### Example 2: Order Matters

Using the same functions,

$$
(g \circ f)(x) = g(f(x)) = g(2x + 1) = (2x + 1)^2.
$$

Since
$$
(f \circ g)(x) \neq (g \circ f)(x),
$$
**composition is not commutative**.

---

### Example 3: Composite with a Rational Function

Let
$$
f(x) = \frac{1}{x}, \qquad g(x) = x - 3.
$$

Then
$$
(f \circ g)(x) = \frac{1}{x - 3}.
$$

---

### Domain of a Composite Function

To find the domain of $f(g(x))$:
1. $x$ must be in the domain of $g$.
2. $g(x)$ must be in the domain of $f$.

---

### Example 4: Domain of a Composite

Using the functions above:

$$
(f \circ g)(x) = \frac{1}{x - 3}
$$

- $g(x) = x - 3$ is defined for all real $x$,
- $f(x) = \frac{1}{x}$ is undefined when its input is 0.

So we require:
$$
x - 3 \neq 0 \Rightarrow x \neq 3.
$$

**Domain:**
$$
(-\infty, 3) \cup (3, \infty).
$$

## 6.2 One-to-One and Inverse Functions

### One-to-One Functions

A function $f$ is **one-to-one** if **different inputs give different outputs**.

Formally:
- If $f(x_1) = f(x_2)$, then $x_1 = x_2$.

This means no two different $x$-values produce the same $y$-value.

---

### Horizontal Line Test

A function is **one-to-one** if it passes the **horizontal line test**:
- Every horizontal line intersects the graph of $f$ at **most one point**.

If a horizontal line intersects the graph more than once, the function is **not** one-to-one.

Below is a short youtube video to introduce you to these topics


```python
YouTubeVideo("6u9Qh1Vi_Qw", width=560, height=315)  # video ID only
```



### Example 1: One-to-One Function

$$
f(x) = 2x + 3
$$

- This is a linear function with nonzero slope.
- Every horizontal line intersects the graph exactly once.
- Therefore, $f(x)$ is **one-to-one**.


### Example 2: Not One-to-One

$$
f(x) = x^2
$$

- This is a parabola opening upward.
- A horizontal line (for example, $y = 4$) intersects the graph at **two points**.
- Therefore, $f(x)$ is **not one-to-one** on all real numbers.

# Exponential Functions 

This module brings together several important families of functions used throughout mathematics, engineering, and the sciences. We will:

- Explore **exponential functions** and the special number \( e \).

You will use algebraic skills and graphing intuition to compare different types of functions and interpret their behavior.

An **exponential function** has the form

$$
f(x) = a \cdot b^x,
$$

where:
- $a$ is a nonzero constant called the **initial value**,
- $b$ is a positive constant with $b \neq 1$, called the **base**.

> **Key idea:** In an exponential function, the variable $x$ appears **in the exponent**, not as a power of $x$.

### Evaluating Exponential Functions

To evaluate $f(x) = a b^x$:

1. Substitute the given value of $x$,
2. Evaluate the exponent,
3. Multiply by $a$.

#### Example 1
Evaluate $f(x) = 3 \cdot 2^x$ at $x = 4$.

$$
f(4) = 3 \cdot 2^4 = 3 \cdot 16 = 48.
$$

#### Example 2
Evaluate $f(x) = 5 \cdot (0.5)^x$ at $x = 2$.

$$
f(2) = 5 \cdot (0.5)^2 = 5 \cdot 0.25 = 1.25.
$$

### Graphing Exponential Functions

The shape of the graph depends on the value of the base $b$.

- If $b > 1$: **exponential growth**
  - The function increases as $x$ increases.
- If $0 < b < 1$: **exponential decay**
  - The function decreases as $x$ increases.

#### Key Graph Features
- The graph always passes through $(0, a)$ because
  $$
  f(0) = a b^0 = a.
  $$
- The x-axis, $y = 0$, is a **horizontal asymptote** when $a > 0$.
- The graph never crosses the x-axis when $a > 0$.

#### Example 3 (Growth)
For $f(x) = 2^x$:
- $b = 2 > 1$, so the function shows **exponential growth**.
- Sample points:
  - $f(0) = 1$
  - $f(1) = 2$
  - $f(2) = 4$

```python
# Define x range
x = np.linspace(-3, 4, 400)

# Example 3: Exponential Growth
y_growth = 2 ** x
plt.figure()
plt.plot(x, y_growth)
plt.axhline(0)
plt.axvline(0)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Example 3: Exponential Growth  f(x) = 2^x")
plt.show()
```



#### Example 4 (Decay)
For $f(x) = (0.5)^x$:
- $0 < b = 0.5 < 1$, so the function shows **exponential decay**.
- Sample points:
  - $f(0) = 1$
  - $f(1) = 0.5$
  - $f(2) = 0.25$

---

```python
# Define x range
x = np.linspace(-3, 4, 400)

# Example 4: Exponential Decay
y_decay = (0.5) ** x
plt.figure()
plt.plot(x, y_decay)
plt.axhline(0)
plt.axvline(0)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Example 4: Exponential Decay  f(x) = (0.5)^x")
plt.show()
```

## Real World Example (Population Growth → Sigmoid → Data Science)

The exponential model

$$
P(t) = 500 \cdot 2^t
$$

describes rapid growth, but it assumes the population can grow **without limit**. In real systems, growth eventually slows because of limited resources.

To model growth that starts fast and then levels off, we use the **sigmoid (logistic) function**, which is built from **exponentials**.


### The Sigmoid Function

The sigmoid function is defined as

$$
S(x) = \frac{1}{1 + e^{-x}}.
$$

This function is exponential because it contains the exponential term $e^{-x}$.

Key features:
- Outputs are always between **0 and 1**
- Growth is fast in the middle and slow at the ends
- The graph has an **S-shape**

Because of this, the sigmoid function is ideal for modeling **probabilities**.


### How This Is Used in Logistic Regression

In **logistic regression**, we use the sigmoid function to convert a linear expression into a probability.

The logistic regression model is

$$
p(x) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 x)}}.
$$

Here:
- $x$ is an input variable,
- $p(x)$ is the probability of an outcome,
- $\beta_0$ and $\beta_1$ are numbers learned from data.

---

### Simple Data Science Example

Suppose we want to predict whether a student **passes** an exam based on hours studied.

Let:
- $x$ = hours studied,
- $p(x)$ = probability of passing.

A logistic regression model might be

$$
p(x) = \frac{1}{1 + e^{-( -4 + 0.8x )}}.
$$

If a student studies $x = 5$ hours,

$$
p(5) = \frac{1}{1 + e^{0}} = 0.5.
$$

**Interpretation:**  
The model predicts a **50% chance** that the student passes.

---

> **Key connection:**  
> Exponential functions lead directly to the sigmoid function, and the sigmoid function is what makes logistic regression work in data science.

```python
def plot_exponential(a=1.0, b=2.0):
    """
    Plot an exponential function f(x) = a * b^x.
    """
    if b <= 0 or b == 1:
        b = 2.0  # enforce valid base
    
    x = np.linspace(-3, 3, 400)
    y = a * (b ** x)
    
    plt.figure(figsize=(6, 4))
    plt.axhline(0, linewidth=1)
    plt.axvline(0, linewidth=1)
    
    plt.plot(x, y, label=rf"$f(x) = {a:.1f} \cdot {b:.2f}^x$")
    
    # Horizontal asymptote y = 0 (if a > 0)
    plt.axhline(0, linestyle='--', label="y = 0 (asymptote)")
    
    plt.ylim(-10, 10)
    plt.legend()
    plt.grid(True)
    plt.title("Exponential Function")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.show()

interact(
    plot_exponential,
    a=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='a'),
    b=widgets.FloatSlider(value=2.0, min=0.2, max=5.0, step=0.2, description='b'),
);
```

---

### Key Takeaways
- Exponential functions grow or decay by **multiplication**.
- The variable $x$ is always in the **exponent**.
- The base $b$ controls whether the function grows or decays.
- The initial value $a$ gives the y-intercept.
