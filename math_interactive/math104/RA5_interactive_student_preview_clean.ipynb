## 1. From Implicit Differentiation to Related Rates

In implicit differentiation we differentiated with respect to $x$:

$$\frac{d}{dx}(y^2)=2y\frac{dy}{dx}$$

In related rates, **every** quantity is a function of time $t$, so we differentiate with respect to $t$:

$$\frac{d}{dt}(x^2)=2x\frac{dx}{dt} \qquad \frac{d}{dt}(r^3)=3r^2\frac{dr}{dt} \qquad \frac{d}{dt}(xy)=x\frac{dy}{dt}+y\frac{dx}{dt}$$

The Chain Rule attaches a rate such as $\frac{dx}{dt}$ or $\frac{dr}{dt}$ to every variable.

#### Reading the notation

- $\frac{dr}{dt}$ = how fast the radius changes (for example, cm per second).
- $\frac{dA}{dt}$ = how fast the area changes (for example, cm² per second).
- A **positive** rate means the quantity is **increasing**; a **negative** rate means it is **decreasing**.

# General Procedure

Whenever you see a related rates problem:

1. **Draw a picture** and label the quantities that change with variables.
2. **Write down** the given rates and the rate you want to find.
3. **Write an equation** that connects the variables.
4. **Differentiate both sides with respect to $t$** (use the Chain Rule).
5. **Substitute** the known values — only **after** differentiating.
6. **Solve** for the unknown rate and include units.

### Example 1 (Circle)

A pebble is dropped into a pond, making a circular ripple. The radius grows at $2$ cm/s. How fast is the area growing when the radius is $10$ cm?

**Equation:**

$$A=\pi r^2$$

**Differentiate with respect to $t$:**

$$\frac{dA}{dt}=2\pi r\frac{dr}{dt}$$

**Substitute** $r=10$ and $\frac{dr}{dt}=2$:

$$\frac{dA}{dt}=2\pi(10)(2)=40\pi \approx 125.66 \text{ cm}^2/\text{s}$$

**Explanation**

We substituted $r=10$ **after** differentiating. If we had put $r=10$ in first, $A$ would be a constant and its derivative would be $0$, which is wrong.

### Example 2 (Square)

The side length $s$ of a square grows at $4$ cm/s. How fast is the area growing when $s=3$ cm?

**Equation:**

$$A=s^2$$

**Differentiate with respect to $t$:**

$$\frac{dA}{dt}=2s\frac{ds}{dt}$$

**Substitute** $s=3$ and $\frac{ds}{dt}=4$:

$$\frac{dA}{dt}=2(3)(4)=24 \text{ cm}^2/\text{s}$$

**Explanation**

The rate $\frac{ds}{dt}$ appears because $s$ depends on $t$, exactly like $\frac{dy}{dx}$ appeared in implicit differentiation.

### Example 3 (Sphere)

Air is pumped into a spherical balloon so that its radius grows at $2$ cm/s. How fast is the volume increasing when $r=3$ cm?

**Equation:**

$$V=\frac{4}{3}\pi r^3$$

**Differentiate with respect to $t$:**

$$\frac{dV}{dt}=4\pi r^2\frac{dr}{dt}$$

**Substitute** $r=3$ and $\frac{dr}{dt}=2$:

$$\frac{dV}{dt}=4\pi(3)^2(2)=72\pi \approx 226.19 \text{ cm}^3/\text{s}$$

**Explanation**

The Power Rule gives $\frac{4}{3}\cdot 3r^2 = 4r^2$, and the Chain Rule adds $\frac{dr}{dt}$.

### Example 4 (Sliding Ladder)

A $10$ ft ladder leans against a wall. The bottom slides away from the wall at $1$ ft/s. How fast is the top sliding down when the bottom is $6$ ft from the wall?

Let $x$ = distance from wall to bottom, $y$ = height of the top.

**Equation (Pythagorean Theorem):**

$$x^2+y^2=100$$

**Differentiate with respect to $t$:**

$$2x\frac{dx}{dt}+2y\frac{dy}{dt}=0 \quad\Longrightarrow\quad \frac{dy}{dt}=-\frac{x}{y}\frac{dx}{dt}$$

**Find $y$** when $x=6$: $\;y=\sqrt{100-36}=8$.

**Substitute** $x=6$, $y=8$, $\frac{dx}{dt}=1$:

$$\frac{dy}{dt}=-\frac{6}{8}(1)=-\frac{3}{4} \text{ ft/s}$$

**Explanation**

The answer is **negative** because the top of the ladder is moving **down** (the height $y$ is decreasing). Notice this is the same equation as the circle $x^2+y^2=25$ from implicit differentiation, now differentiated with respect to $t$.

### Example 5 (Two Moving Cars)

Car A drives north from an intersection at $20$ mph and car B drives east from the same intersection at $30$ mph. How fast is the distance between them changing when car A is $6$ miles north and car B is $8$ miles east?

Let $y$ = distance of car A, $x$ = distance of car B, $z$ = distance between the cars.

**Equation:**

$$x^2+y^2=z^2$$

**Differentiate with respect to $t$:**

$$2x\frac{dx}{dt}+2y\frac{dy}{dt}=2z\frac{dz}{dt} \quad\Longrightarrow\quad \frac{dz}{dt}=\frac{x\frac{dx}{dt}+y\frac{dy}{dt}}{z}$$

**Find $z$:** $\;z=\sqrt{8^2+6^2}=10$.

**Substitute** $x=8$, $\frac{dx}{dt}=30$, $y=6$, $\frac{dy}{dt}=20$:

$$\frac{dz}{dt}=\frac{8(30)+6(20)}{10}=\frac{240+120}{10}=36 \text{ mph}$$

**Explanation**

Here all three sides of the triangle are changing, so all three have a rate.

# Common Mistakes

- Substituting numbers **before** differentiating.
- Forgetting the Chain Rule (leaving out $\frac{dr}{dt}$, $\frac{dx}{dt}$, …).
- Using the wrong sign: a quantity that is **decreasing** has a **negative** rate.
- Forgetting to find a missing value (like $y$ in the ladder problem) before substituting.
- Leaving off units, or using the wrong units (area rates are in units² per time).

# Real-World Applications

Related rates appear in:

- Filling or draining tanks and reservoirs
- Inflating balloons and airbags
- Spreading oil spills and wildfires
- Air-traffic control and radar tracking
- Shadows and moving light sources
- Blood flow and medical dosing
- Economics (how cost changes as production changes over time)
