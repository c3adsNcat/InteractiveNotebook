## 1. Angles in Standard Position

An **angle** is formed by rotating a ray around its endpoint, the **vertex**. The starting ray is the **initial side** and the ending ray is the **terminal side**.

An angle is in **standard position** when its vertex is at the origin and its initial side lies on the positive $x$-axis.

- Rotating **counterclockwise** gives a **positive** angle.
- Rotating **clockwise** gives a **negative** angle.



#### 🖼️ Image description
Two coordinate planes, side by side. In each, the vertex is at the origin and the **initial side** is a thick black ray along the positive $x$-axis.
- **Left:** a blue curved arrow turns **counterclockwise** 120° to a blue **terminal side** in the second quadrant (upper left). This is a **positive** angle.
- **Right:** an orange curved arrow turns **clockwise** 45° to an orange **terminal side** in the fourth quadrant (lower right). This is a **negative** angle, −45°.

### Example 1 (Coterminal Angles)

Angles with the **same terminal side** are called **coterminal**. To find one, add or subtract a full rotation, $360^\circ$ (or $2\pi$ radians).

For $\theta = 30^\circ$:

$$30^\circ + 360^\circ = 390^\circ \qquad\qquad 30^\circ - 360^\circ = -330^\circ$$

So $390^\circ$ and $-330^\circ$ are both coterminal with $30^\circ$.

**Explanation**

One full rotation brings the terminal side back to the same place, so adding or subtracting $360^\circ$ any number of times never changes where the angle ends.

## 2. Degrees and Radians

There are two common units for angles:

- **Degrees:** one full rotation is $360^\circ$.
- **Radians:** an angle of **1 radian** cuts off an arc whose length equals the radius. One full rotation is $2\pi$ radians.

Since a full rotation is both $360^\circ$ and $2\pi$ radians,

$$180^\circ = \pi \text{ radians}$$

$$\text{degrees} \to \text{radians: multiply by } \frac{\pi}{180^\circ} \qquad\qquad \text{radians} \to \text{degrees: multiply by } \frac{180^\circ}{\pi}$$



#### 🖼️ Image description
A circle centered at the origin with 16 points marked on it at common angles. Next to each point the angle is written **in degrees** (black, outside the circle) and **in radians** (orange, inside the circle), for example: 0° = 0, 30° = π/6, 45° = π/4, 60° = π/3, 90° = π/2, 180° = π, 270° = 3π/2, and 330° = 11π/6. Moving counterclockwise around the circle, the angles increase.

### Example 2 (Converting)

**Degrees to radians:** $\;60^\circ \cdot \dfrac{\pi}{180^\circ} = \dfrac{60\pi}{180} = \dfrac{\pi}{3}$ radians

**Radians to degrees:** $\;\dfrac{3\pi}{4} \cdot \dfrac{180^\circ}{\pi} = \dfrac{540^\circ}{4} = 135^\circ$

**One radian in degrees:** $\;1 \cdot \dfrac{180^\circ}{\pi} \approx 57.3^\circ$

**Explanation**

Choose the conversion factor so the unit you **don't** want cancels: to remove degrees, put $180^\circ$ in the denominator.

## 3. Arc Length and Area of a Sector

For a circle of radius $r$ and a central angle $\theta$ **measured in radians**:

$$\text{Arc length: } s = r\theta \qquad\qquad \text{Area of a sector: } A = \tfrac{1}{2} r^2 \theta$$



#### 🖼️ Image description
A circle with a central angle $\theta$ formed by two radii, each labeled $r$. The part of the circle between the two radii is highlighted in **orange** and labeled **arc length s**. The pie-slice region between the radii is shaded **blue** and labeled **sector area A**. Below the circle are the formulas $s = r\theta$ and $A = \tfrac{1}{2}r^2\theta$, with the reminder that $\theta$ must be in **radians**.

### Example 3 (Arc Length and Sector Area)

A circle has radius $r = 4$ m and central angle $\theta = \dfrac{\pi}{2}$.

$$s = r\theta = 4 \cdot \frac{\pi}{2} = 2\pi \approx 6.28 \text{ m}$$

$$A = \tfrac{1}{2} r^2 \theta = \tfrac{1}{2}(4)^2 \cdot \frac{\pi}{2} = 4\pi \approx 12.57 \text{ m}^2$$

**Explanation**

If the angle is given in degrees, **convert it to radians first**. For example, $90^\circ$ must be written as $\dfrac{\pi}{2}$ before using these formulas.

## 4. Linear Speed and Angular Speed

When an object moves around a circle of radius $r$:

- **Angular speed** $\omega$ (omega) is how fast the **angle** changes: $\;\omega = \dfrac{\theta}{t}$ (radians per unit of time).
- **Linear speed** $v$ is how fast the **distance along the circle** changes: $\;v = \dfrac{s}{t}$.

Because $s = r\theta$, the two are connected by

$$v = r\omega$$

### Example 4 (A Merry-Go-Round)

A child sits $5$ ft from the center of a merry-go-round that turns at $\omega = 2$ radians per second.

$$v = r\omega = 5 \cdot 2 = 10 \text{ feet per second}$$

**Explanation**

If the angular speed is given in **revolutions** per minute, convert first using $1 \text{ revolution} = 2\pi$ radians. Radians are what make $v = r\omega$ work without extra conversion factors.

## 5. The Six Trigonometric Functions

In a right triangle, choose one of the acute angles and call it $\theta$. The sides are named **relative to $\theta$**:



#### 🖼️ Image description
A right triangle with the right angle (small square) at the lower right corner and the angle $\theta$ at the lower left corner.
- The **hypotenuse** (blue) is the longest side, across from the right angle.
- The **opposite** side (orange) is the vertical side across from $\theta$.
- The **adjacent** side (green) is the horizontal side next to $\theta$ that is not the hypotenuse.

A box to the right shows the memory aid **SOH-CAH-TOA**: sin θ = opposite/hypotenuse, cos θ = adjacent/hypotenuse, tan θ = opposite/adjacent.

The six trigonometric functions of $\theta$ are:

| Function | Ratio | Reciprocal function | Ratio |
|---|---|---|---|
| $\sin\theta$ | $\dfrac{\text{opposite}}{\text{hypotenuse}}$ | $\csc\theta = \dfrac{1}{\sin\theta}$ | $\dfrac{\text{hypotenuse}}{\text{opposite}}$ |
| $\cos\theta$ | $\dfrac{\text{adjacent}}{\text{hypotenuse}}$ | $\sec\theta = \dfrac{1}{\cos\theta}$ | $\dfrac{\text{hypotenuse}}{\text{adjacent}}$ |
| $\tan\theta$ | $\dfrac{\text{opposite}}{\text{adjacent}}$ | $\cot\theta = \dfrac{1}{\tan\theta}$ | $\dfrac{\text{adjacent}}{\text{opposite}}$ |

### Example 5 (Finding All Six Values)

In a right triangle, the side **opposite** $\theta$ is $8$, the side **adjacent** to $\theta$ is $15$, and the **hypotenuse** is $17$.

$$\sin\theta = \frac{8}{17} \qquad \cos\theta = \frac{15}{17} \qquad \tan\theta = \frac{8}{15}$$

$$\csc\theta = \frac{17}{8} \qquad \sec\theta = \frac{17}{15} \qquad \cot\theta = \frac{15}{8}$$

**Explanation**

Find $\sin$, $\cos$, and $\tan$ first with SOH-CAH-TOA. The other three are their **reciprocals** (flip the fraction).

## 6. Fundamental Identities and Complementary Angles

These identities are true for every acute angle $\theta$:

- **Quotient identity:** $\;\tan\theta = \dfrac{\sin\theta}{\cos\theta}$
- **Pythagorean identity:** $\;\sin^2\theta + \cos^2\theta = 1$

Two acute angles are **complementary** if they add up to $90^\circ$. In a right triangle, the two acute angles are complementary, and the side opposite one is adjacent to the other. This gives the **Complementary Angle Theorem**:

$$\sin\theta = \cos(90^\circ - \theta) \qquad\qquad \cos\theta = \sin(90^\circ - \theta)$$

### Example 7

$$\sin 30^\circ = \cos 60^\circ = \tfrac{1}{2} \qquad\qquad \cos 25^\circ = \sin 65^\circ$$

Check the Pythagorean identity with Example 6: $\;\left(\tfrac{4}{5}\right)^2 + \left(\tfrac{3}{5}\right)^2 = \tfrac{16}{25} + \tfrac{9}{25} = 1$ ✓

### 📈 Visualization: Sine and Cosine as the Angle Grows

Run the cell below. It plots $\sin\theta$ and $\cos\theta$ for angles from $0^\circ$ to $90^\circ$. Look for:

- where the two curves **cross** (at $45^\circ$, where opposite = adjacent);
- the **mirror symmetry** around $45^\circ$: the value of $\sin$ at $\theta$ equals the value of $\cos$ at $90^\circ - \theta$.

```python
import numpy as np
import matplotlib.pyplot as plt

degrees = np.linspace(0, 90, 200)
radians = np.radians(degrees)          # Python's trig functions need RADIANS

fig, ax = plt.subplots(figsize=(7.5, 4.2))
ax.plot(degrees, np.sin(radians), color="#2a78d6", lw=2, label="sin θ")
ax.plot(degrees, np.cos(radians), color="#eb6834", lw=2, label="cos θ")
ax.axvline(45, color="gray", ls=":", lw=1)
ax.text(46, 0.05, "45°: sin θ = cos θ", color="gray")
ax.text(88, 0.94, "sin θ", color="#2a78d6", ha="right", weight="bold")
ax.text(2, 0.94, "cos θ", color="#eb6834", weight="bold")
ax.set_xlabel("Angle θ (degrees)"); ax.set_ylabel("Value of the ratio")
ax.set_xticks(range(0, 91, 15)); ax.set_ylim(0, 1.05)
ax.set_title("Sine rises and cosine falls as θ goes from 0° to 90°", loc="left")
ax.legend(frameon=False, loc="center right")
ax.grid(alpha=0.3)
plt.show()
```

## 7. Application: Finding a Height

Trigonometry lets us measure things we can't reach. The **angle of elevation** is the angle between the horizontal ground and your line of sight **up** to an object.

### Example 8 (How Tall Is the Tree?)

You stand $40$ feet from the base of a tree. The angle of elevation to the top of the tree is $50^\circ$. The height $h$ is **opposite** the angle and the $40$ ft distance is **adjacent**, so use tangent:

$$\tan 50^\circ = \frac{h}{40} \quad\Longrightarrow\quad h = 40\tan 50^\circ \approx 40(1.1918) \approx 47.67 \text{ feet}$$

**Explanation**

Calculators must be in **degree mode** for $\tan 50^\circ$. In Python, `math.tan` and `np.tan` always expect **radians**, so convert first: `math.tan(math.radians(50))`.

# Common Mistakes

- Using **degrees** in $s = r\theta$, $A = \tfrac12 r^2\theta$, or $v = r\omega$. These formulas need **radians**.
- Flipping a conversion factor: degrees → radians uses $\dfrac{\pi}{180^\circ}$, not $\dfrac{180^\circ}{\pi}$.
- Naming sides from the wrong angle: "opposite" and "adjacent" depend on **which** angle is $\theta$.
- Mixing up reciprocals: $\csc\theta = \dfrac{1}{\sin\theta}$ (not $\dfrac{1}{\cos\theta}$).
- Forgetting that a calculator must be in **degree mode**, and that Python's `math.sin`, `math.cos`, `math.tan` always use **radians**.

# Connecting to Data Science

Angles and trigonometry show up in data science more often than you might expect.

### 1. Computers think in radians

Almost every programming tool (Python's `math`, NumPy, Excel, and others) computes trig functions in **radians**. A very common bug is forgetting to convert:

- `np.sin(30)` gives about **−0.988** (the sine of 30 *radians*).
- `np.sin(np.radians(30))` gives **0.5** (the sine of 30 *degrees*). ✓

Knowing $180^\circ = \pi$ radians helps you spot and fix this kind of bug.

### 2. Data that goes in circles

Some data is **cyclical**: hours of the day, days of the week, months, wind direction, compass headings. If we feed the hour as a plain number to a machine-learning model, it thinks 11 PM (23) and midnight (0) are far apart, even though they're only one hour apart. Data scientists fix this by turning each value into an **angle** and using its **cosine and sine**:

$$\theta = 2\pi \cdot \frac{\text{hour}}{24} \qquad\longrightarrow\qquad (\cos\theta,\ \sin\theta)$$



#### 🖼️ Image description
**Left panel ("Before"):** the 24 hours of the day are shown as dots on a straight number line from 0 to 23. The dots for midnight (0) and 11 PM (23) are highlighted in orange at opposite ends, with an arrow labeled "look 23 apart."
**Right panel ("After"):** the same 24 hours are placed evenly around a circle using the angle $\theta = 2\pi \cdot \text{hour}/24$ and the point $(\cos\theta, \sin\theta)$. Hours 0, 6, 12, and 18 are labeled at the right, top, left, and bottom. Now the orange dots for 0 and 23 sit **right next to each other**, which matches reality.

### 3. Measuring similarity with angles

Search engines and recommendation systems (like Netflix or Spotify) represent movies, songs, or documents as arrows (vectors) of numbers. The **angle** between two arrows measures how similar they are: a small angle (large $\cos\theta$) means "very similar." This measure is called **cosine similarity**.

### 4. Measuring the world from images

Drones, phone apps, and self-driving cars estimate heights and distances from camera angles, exactly like the tree and drone problems in this assignment, using $h = d\tan\theta$.

Run the cell below to see the radians bug, and the cyclical encoding of a few hours, for yourself.

```python
import numpy as np

print("np.sin(30)             =", round(np.sin(30), 4), "   <- sine of 30 RADIANS (wrong for 30°)")
print("np.sin(np.radians(30)) =", round(np.sin(np.radians(30)), 4), "      <- sine of 30 DEGREES ✓")
print()

# Cyclical encoding: hours become points on a circle
for hour in [0, 6, 12, 18, 23]:
    theta = 2 * np.pi * hour / 24
    print(f"hour {hour:2d}:  angle = {np.degrees(theta):5.1f}°   point = ({np.cos(theta):6.3f}, {np.sin(theta):6.3f})")
```
